### Exercise 03: Create a null vector of size 10
**Difficulty:** `★☆☆`  
**Tags:** `Array Creation`

#### 💡 Intuition & Concept
`np.zeros` allocates a contiguous block of memory initialized to zero. By default, it uses `np.float64` (double-precision float).

#### ⚠️ Key Takeaway & Gotchas
If you need integer zeros, always specify `dtype=int` or `dtype=np.int32` to avoid unnecessary floating-point storage.

### Exercise 04: Find the memory size of any array
**Difficulty:** `★☆☆`  
**Tags:** `Memory, Attributes`

#### 💡 Intuition & Concept
Every NumPy array maintains metadata. The total byte size can be queried directly via the `nbytes` attribute, or calculated as the number of elements (`arr.size`) multiplied by the size of each element in bytes (`arr.itemsize`).

#### ⚠️ Key Takeaway & Gotchas
`sys.getsizeof(arr)` in Python returns the size of the Python wrapper object, whereas `arr.nbytes` returns the actual memory buffer containing the raw array elements.


```python
Z = np.zeros((10, 10))
print(f"Total elements (size): {Z.size}")
print(f"Bytes per element (itemsize): {Z.itemsize} bytes")
print(f"Total memory buffer (nbytes): {Z.nbytes} bytes")
```

### Exercise 07: Create a vector with values ranging from 10 to 49
**Difficulty:** `★☆☆`  
**Tags:** `Array Creation, Range`

#### 💡 Intuition & Concept
`np.arange(start, stop)` creates an evenly spaced integer sequence over the half-open interval `[start, stop)`. To include 49, stop must be set to 50.

#### ⚠️ Key Takeaway & Gotchas
For non-integer steps, prefer `np.linspace` instead of `np.arange` to prevent floating-point rounding accumulation.


```python
Z = np.arange(10, 50)
print(Z)
```

### Exercise 08: Reverse a vector (first element becomes last)
**Difficulty:** `★☆☆`  
**Tags:** `Slicing, Views`

#### 💡 Intuition & Concept
Using slice stride `-1` (`Z[::-1]`) reverses the array without copying data. It creates a view with negative strides, running in $O(1)$ time.

#### ⚠️ Key Takeaway & Gotchas
Because `Z[::-1]` is a view, modifying elements in the reversed view will modify the original array! If you want an independent copy, call `.copy()`.


```python
Z = np.arange(10)
reversed_Z = Z[::-1]
print("Original:", Z)
print("Reversed:", reversed_Z)
```

#### 🔄 Alternative Approach / Benchmark
Using `np.flip` provides a readable alternative that supports multi-dimensional reversal:

```python
# Using np.flip (also returns a view):
flipped = np.flip(Z)
print("np.flip:", flipped)
```

### Exercise 09: Create a 3x3 matrix with values ranging from 0 to 8
**Difficulty:** `★☆☆`  
**Tags:** `Reshaping, Array Creation`

#### 💡 Intuition & Concept
`np.arange(9)` creates 9 elements, and `.reshape(3, 3)` reorganizes the shape without altering or copying memory data when contiguous.

#### ⚠️ Key Takeaway & Gotchas
The product of the target shape dimensions must exactly equal the total number of elements ($3 \times 3 = 9$).


```python
Z = np.arange(9).reshape(3, 3)
print(Z)
```

### Exercise 10: Find indices of non-zero elements from [1,2,0,0,4,0]
**Difficulty:** `★☆☆`  
**Tags:** `Searching, Indexing`

#### 💡 Intuition & Concept
`np.nonzero(arr)` returns a tuple of arrays, one for each dimension, containing the indices where elements are non-zero.

#### ⚠️ Key Takeaway & Gotchas
`np.nonzero(arr)[0]` extracts the 1D index array directly. Alternatively, `np.flatnonzero(arr)` returns flattened indices.


```python
arr = np.array([1, 2, 0, 0, 4, 0])
nz_indices = np.nonzero(arr)
print("Non-zero indices:", nz_indices[0])
```

#### 🔄 Alternative Approach / Benchmark
Using `np.flatnonzero`:

```python
# Alternative using flatnonzero:
fnz = np.flatnonzero(arr)
print("flatnonzero:", fnz)
```

### Exercise 11: Create a 3x3 identity matrix
**Difficulty:** `★☆☆`  
**Tags:** `Linear Algebra, Matrix Creation`

#### 💡 Intuition & Concept
An identity matrix has ones on the main diagonal and zeros elsewhere. `np.eye(N)` creates an $N \times N$ identity matrix.

#### ⚠️ Key Takeaway & Gotchas
`np.eye` allows non-square shapes and shifting the diagonal via parameter `k`, while `np.identity` creates only square matrices.


```python
Z = np.eye(3)
print(Z)
```

### Exercise 12: Create a 3x3x3 array with random values
**Difficulty:** `★☆☆`  
**Tags:** `Random, 3D Arrays`

#### 💡 Intuition & Concept
In modern NumPy (v1.17+ and NumPy 2.x), the recommended way to generate random numbers is via `np.random.default_rng()`.

#### ⚠️ Key Takeaway & Gotchas
Avoid legacy `np.random.random()` or `np.random.rand()` in modern codebases, as `default_rng()` provides a faster, higher-quality PCG64 PRNG.


```python
rng = np.random.default_rng(seed=42)
Z = rng.random((3, 3, 3))
print("Shape:", Z.shape)
print(Z)
```

### Exercise 13: Create a 10x10 array with random values and find min and max
**Difficulty:** `★☆☆`  
**Tags:** `Statistics, Reductions`

#### 💡 Intuition & Concept
`arr.min()` and `arr.max()` compute global extremes across the entire array in a fast vectorized C loop.

#### ⚠️ Key Takeaway & Gotchas
You can also specify `axis=0` (column-wise) or `axis=1` (row-wise) if reductions along specific dimensions are required.


```python
rng = np.random.default_rng(seed=123)
Z = rng.random((10, 10))
print(f"Minimum value: {Z.min():.4f}")
print(f"Maximum value: {Z.max():.4f}")
```

### Exercise 14: Create a random vector of size 30 and find the mean value
**Difficulty:** `★☆☆`  
**Tags:** `Statistics, Mean`

#### 💡 Intuition & Concept
`Z.mean()` computes the arithmetic mean. For uniform random variables in $[0, 1)$, the theoretical mean converges to $0.5$.

#### ⚠️ Key Takeaway & Gotchas
For arrays containing potential `np.nan` values, `np.mean` will return `nan`. In those cases, use `np.nanmean`.


```python
rng = np.random.default_rng(seed=7)
Z = rng.random(30)
print(f"Mean value: {Z.mean():.4f}")
```

### Exercise 15: Create a 2D array with 1 on the border and 0 inside
**Difficulty:** `★☆☆`  
**Tags:** `Indexing, Slicing`

#### 💡 Intuition & Concept
Start with an array of ones using `np.ones((rows, cols))`, then use multi-slice `Z[1:-1, 1:-1] = 0` to set the interior region to zero.

#### ⚠️ Key Takeaway & Gotchas
Negative indexing `-1` refers to the last element (exclusive in a slice), neatly skipping the outer boundaries on all 4 sides.


```python
Z = np.ones((6, 6), dtype=int)
Z[1:-1, 1:-1] = 0
print(Z)
```
